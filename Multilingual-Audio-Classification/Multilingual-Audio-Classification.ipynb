```python
!pip install transformers datasets

from transformers import pipeline

whisper_fleurs = pipeline("audio-classification", model="sanchit-gandhi/whisper-medium-fleurs-lang-id")

import librosa
```

```python
# Italian Language Identification

# List of audio file paths
audio_file_paths = ['/content/1 IT.wav', '/content/2 IT.wav', '/content/3 IT.wav', '/content/4 IT.wav', '/content/5 IT.wav', '/content/6 IT.wav', '/content/7 IT.wav', '/content/8 IT.wav', '/content/9 IT.wav', '/content/10 IT.wav']

# Process each audio file
for audio_file_path in audio_file_paths:
    # Load the .wav file
    audio, sampling_rate = librosa.load(audio_file_path, sr=16000)

    # Perform language identification
    results = whisper_fleurs(audio)

    # Print the results
    print(f"Results for {audio_file_path}:")
    print(results)
```
